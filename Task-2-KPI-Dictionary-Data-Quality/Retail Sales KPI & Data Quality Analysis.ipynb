# Import Libraries

```python
# Import pandas for data analysis
import pandas as pd

# Import numpy for numerical calculations
import numpy as np

# Import os for checking files
import os

# Display all columns in output
pd.set_option("display.max_columns", None)

# Display maximum 100 rows
pd.set_option("display.max_rows", 100)

# Print confirmation message
print("All libraries imported successfully.")
```

# Load CSV Files

```python
# Load the retail orders CSV file
orders = pd.read_csv("retail-orders-raw.csv")

# Load the retail data dictionary CSV file
data_dictionary = pd.read_csv("retail-data-dictionary.csv")

# Print confirmation
print("Both CSV files loaded successfully.")
```

# Display Data

```python
# Display first five rows of retail orders data
print("Retail Orders Data:")
display(orders.head())

# Display first five rows of data dictionary
print("Retail Data Dictionary:")
display(data_dictionary.head())
```

# Dataset Shape

```python
# Display the number of rows and columns
print("Orders Dataset Shape:", orders.shape)

# Display data dictionary shape
print("Data Dictionary Shape:", data_dictionary.shape)
```

# Column Names

```python
# Display all column names
print("Orders Dataset Columns:")
print(orders.columns.tolist())

print("\nData Dictionary Columns:")
print(data_dictionary.columns.tolist())
```

# Dataset Information

```python
# Display complete information about the orders dataset
orders.info()
```

# Missing Values

```python
# Count missing values in each column
missing_values = orders.isnull().sum()

# Create missing value report
missing_report = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values
})

# Display missing value report
display(missing_report)
```

# Duplicate Values

```python
# Count duplicate rows
duplicate_count = orders.duplicated().sum()

# Print duplicate count
print("Total Duplicate Rows:", duplicate_count)
```

# Unique Values

```python
# Count unique values in each column
unique_values = orders.nunique()

# Create unique values report
unique_report = pd.DataFrame({
    "Column": unique_values.index,
    "Unique Values": unique_values.values
})

# Display unique values report
display(unique_report)
```

# Numerical Columns

```python
# Select numerical columns
numeric_columns = orders.select_dtypes(include=np.number).columns

# Display numerical column names
print("Numerical Columns:")
print(numeric_columns.tolist())

# Display statistical summary
display(orders[numeric_columns].describe())
```

# Convert Date Columns

```python
# Detect columns containing date or time in their names
date_columns = [
    column for column in orders.columns
    if "date" in column.lower() or "time" in column.lower()
]

# Display detected date columns
print("Detected Date Columns:", date_columns)

# Convert date columns into datetime format
for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

# Display updated data types
if date_columns:
    display(orders[date_columns].dtypes)
else:
    print("No date column found.")
```

# Display Complete Dataset

```python
# Display the complete dataset
display(orders)
```

# Calculate Sales and Discount

```python
# Convert quantity column into numeric values.
# Invalid values such as "two" will become NaN.
orders["quantity_numeric"] = pd.to_numeric(
    orders["quantity"],
    errors="coerce"
)

# Convert unit price column into numeric values.
orders["unit_price_numeric"] = pd.to_numeric(
    orders["unit_price"],
    errors="coerce"
)

# Convert discount percentage into numeric values.
orders["discount_numeric"] = pd.to_numeric(
    orders["discount_pct"],
    errors="coerce"
)

# Calculate gross sales before discount.
orders["gross_sales"] = (
    orders["quantity_numeric"] *
    orders["unit_price_numeric"]
)

# Calculate discount amount.
orders["discount_amount"] = (
    orders["gross_sales"] *
    orders["discount_numeric"] / 100
)

# Calculate net sales after discount.
orders["net_sales"] = (
    orders["gross_sales"] -
    orders["discount_amount"]
)

# Display calculated sales columns.
display(
    orders[
        [
            "order_id",
            "quantity_numeric",
            "unit_price_numeric",
            "discount_numeric",
            "gross_sales",
            "discount_amount",
            "net_sales"
        ]
    ]
)
```

# Calculate KPI Values

```python
# Calculate total gross sales before discount.
total_gross_sales = orders["gross_sales"].sum()

# Calculate total discount amount.
total_discount = orders["discount_amount"].sum()

# Calculate total net sales after discount.
total_net_sales = orders["net_sales"].sum()

# Calculate total valid quantity sold.
total_quantity = orders["quantity_numeric"].sum()

# Calculate total unique orders.
total_orders = orders["order_id"].nunique()

# Calculate average order value.
average_order_value = total_net_sales / total_orders

# Display all KPI values.
print("Total Gross Sales:", total_gross_sales)
print("Total Discount:", total_discount)
print("Total Net Sales:", total_net_sales)
print("Total Quantity:", total_quantity)
print("Total Orders:", total_orders)
print("Average Order Value:", average_order_value)
```

# Create KPI Dictionary

```python
# Create a KPI dictionary according to available columns.
kpi_dictionary = pd.DataFrame({

    # Name of each KPI.
    "KPI Name": [
        "Total Gross Sales",
        "Total Discount",
        "Total Net Sales",
        "Total Quantity",
        "Total Orders",
        "Average Order Value"
    ],

    # Definition of each KPI.
    "Definition": [
        "Total sales before discount",
        "Total discount amount applied",
        "Total sales after discount",
        "Total valid quantity sold",
        "Total number of unique orders",
        "Average net sales per order"
    ],

    # Formula used to calculate each KPI.
    "Formula": [
        "SUM(quantity * unit_price)",
        "SUM(gross_sales * discount_pct / 100)",
        "SUM(gross_sales - discount_amount)",
        "SUM(valid quantity)",
        "COUNT DISTINCT(order_id)",
        "Total Net Sales / Total Orders"
    ],

    # Business meaning of each KPI.
    "Business Meaning": [
        "Measures total sales before discount",
        "Measures discount given to customers",
        "Measures actual sales after discount",
        "Measures total product quantity",
        "Measures order demand",
        "Measures average order value"
    ]
})

# Display the KPI dictionary.
display(kpi_dictionary)
```

# Create KPI Summary

```python
# Create a summary table containing KPI values.
kpi_summary = pd.DataFrame({

    # KPI names.
    "KPI": [
        "Total Gross Sales",
        "Total Discount",
        "Total Net Sales",
        "Total Quantity",
        "Total Orders",
        "Average Order Value"
    ],

    # Calculated KPI values.
    "Value": [
        total_gross_sales,
        total_discount,
        total_net_sales,
        total_quantity,
        total_orders,
        average_order_value
    ]
})

# Display the KPI summary.
display(kpi_summary)
```

# Data Quality Report

```python
# Count total missing values in the dataset.
total_missing_values = orders.isnull().sum().sum()

# Count total duplicate rows.
total_duplicate_rows = orders.duplicated().sum()

# Count negative quantity values.
negative_quantity_values = (
    orders["quantity_numeric"] < 0
).sum()

# Count invalid quantity values.
invalid_quantity_values = (
    orders["quantity_numeric"].isnull()
).sum()

# Count invalid discount values.
invalid_discount_values = (
    (orders["discount_numeric"] < 0) |
    (orders["discount_numeric"] > 100)
).sum()

# Create data quality report.
data_quality_report = pd.DataFrame({

    # Name of each quality check.
    "Check": [
        "Total Rows",
        "Total Columns",
        "Total Missing Values",
        "Duplicate Rows",
        "Negative Quantity Values",
        "Invalid Quantity Values",
        "Invalid Discount Values"
    ],

    # Result of each quality check.
    "Result": [
        orders.shape[0],
        orders.shape[1],
        total_missing_values,
        total_duplicate_rows,
        negative_quantity_values,
        invalid_quantity_values,
        invalid_discount_values
    ],

    # Status of each quality check.
    "Status": [
        "Pass",
        "Pass",
        "Pass" if total_missing_values == 0 else "Review Required",
        "Pass" if total_duplicate_rows == 0 else "Review Required",
        "Pass" if negative_quantity_values == 0 else "Review Required",
        "Pass" if invalid_quantity_values == 0 else "Review Required",
        "Pass" if invalid_discount_values == 0 else "Review Required"
    ]
})

# Display the data quality report.
display(data_quality_report)
```

# Missing Values Report

```python
# Create a detailed missing values report.
missing_values_report = pd.DataFrame({

    # Column names.
    "Column": orders.columns,

    # Number of missing values in each column.
    "Missing Values": orders.isnull().sum().values,

    # Percentage of missing values in each column.
    "Missing Percentage": (
        orders.isnull().sum().values / len(orders)
    ) * 100
})

# Display the missing values report.
display(missing_values_report)
```

# Duplicate Rows Report

```python
# Select all duplicate rows.
duplicate_rows = orders[
    orders.duplicated(keep=False)
]

# Display duplicate rows.
display(duplicate_rows)
```

# Save Data Quality CSV

```python
# Save the data quality report as a CSV file.
data_quality_report.to_csv(
    "data_quality_report.csv",
    index=False
)

# Print confirmation message.
print("data_quality_report.csv created successfully.")
```

# Create Excel Workbook

```python
# Create an Excel workbook with multiple sheets.
with pd.ExcelWriter(
    "KPI_Dictionary_and_Data_Quality.xlsx",
    engine="openpyxl"
) as writer:

    # Save KPI dictionary in Excel.
    kpi_dictionary.to_excel(
        writer,
        sheet_name="KPI Dictionary",
        index=False
    )

    # Save KPI summary in Excel.
    kpi_summary.to_excel(
        writer,
        sheet_name="KPI Summary",
        index=False
    )

    # Save data quality report in Excel.
    data_quality_report.to_excel(
        writer,
        sheet_name="Data Quality Report",
        index=False
    )

    # Save missing values report in Excel.
    missing_values_report.to_excel(
        writer,
        sheet_name="Missing Values",
        index=False
    )

    # Save duplicate rows in Excel.
    duplicate_rows.to_excel(
        writer,
        sheet_name="Duplicate Rows",
        index=False
    )

# Print confirmation message.
print("KPI_Dictionary_and_Data_Quality.xlsx created successfully.")
```
