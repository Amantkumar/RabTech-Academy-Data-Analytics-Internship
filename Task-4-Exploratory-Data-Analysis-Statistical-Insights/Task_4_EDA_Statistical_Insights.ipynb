# Mean, Median and Standard Deviation

```python
# Select numerical columns
numeric_columns = df.select_dtypes(
    include=np.number
).columns

# Calculate mean
mean_values = df[numeric_columns].mean()

# Calculate median
median_values = df[numeric_columns].median()

# Calculate standard deviation
std_values = df[numeric_columns].std()

# Create statistical summary table
statistical_summary = pd.DataFrame({
    "Mean": mean_values,
    "Median": median_values,
    "Standard Deviation": std_values
})

# Display statistical summary
display(statistical_summary)
```

# Quartile Analysis

```python
# Calculate first quartile
Q1 = df[numeric_columns].quantile(0.25)

# Calculate second quartile or median
Q2 = df[numeric_columns].quantile(0.50)

# Calculate third quartile
Q3 = df[numeric_columns].quantile(0.75)

# Create quartile summary
quartile_summary = pd.DataFrame({
    "Q1 - 25%": Q1,
    "Q2 - 50%": Q2,
    "Q3 - 75%": Q3
})

# Display quartile summary
display(quartile_summary)
```

# Histogram Distributions

# Sales Distribution

```python
# Select sales column
sales_column = None

# Identify available sales-related column
for column in [
    "calculated_sales",
    "total_amount",
    "sales",
    "price"
]:
    if column in df.columns:
        sales_column = column
        break

# Display selected sales column
print("Sales Column Used:", sales_column)
```

# Sales by Category

```python
# Check whether category column is available
if "category" in df.columns and sales_column is not None:

    # Group sales by category
    category_sales = (
        df.groupby("category")[sales_column]
        .sum()
        .sort_values(ascending=False)
    )

    # Display category sales
    display(category_sales)

    # Plot category sales
    plt.figure(figsize=(10, 5))

    category_sales.plot(
        kind="bar",
        edgecolor="black"
    )

    plt.title("Sales by Category")
    plt.xlabel("Category")
    plt.ylabel("Sales")
    plt.xticks(rotation=45)
    plt.show()

else:
    print("Category or sales column is not available.")
```

# Sales by Customer Segment

```python
# Check whether customer segment is available
if "customer_segment" in df.columns and sales_column is not None:

    # Group sales by customer segment
    segment_sales = (
        df.groupby("customer_segment")[sales_column]
        .sum()
        .sort_values(ascending=False)
    )

    # Display segment sales
    display(segment_sales)

    # Plot segment sales
    plt.figure(figsize=(8, 5))

    segment_sales.plot(
        kind="bar",
        edgecolor="black"
    )

    plt.title("Sales by Customer Segment")
    plt.xlabel("Customer Segment")
    plt.ylabel("Sales")
    plt.xticks(rotation=45)
    plt.show()

else:
    print("Customer segment or sales column is not available.")
```

# Monthly Sales Trend

```python
# Monthly Sales Trend

if "order_date" in df.columns and sales_column is not None:

    df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")

    monthly_sales = (
        df.groupby(df["order_date"].dt.to_period("M"))[sales_column]
        .sum()
    )

    display(monthly_sales)

    plt.figure(figsize=(10, 5))

    monthly_sales.plot(
        kind="line",
        marker="o"
    )

    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Sales")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

else:
    print("Order date column or sales column is not available.")
```

# Fix Delivery Time Column

```python
# Fix Delivery Time Column

import pandas as pd
import numpy as np

# Convert delivery time values into numeric days
def convert_delivery_days(value):
    try:
        value = str(value)

        # If value is like 1970-01-01 00:00:00.000000004
        if "." in value:
            number = value.split(".")[-1]
            number = number.replace(" ", "")
            return float(number)

        # If value is already numeric
        return float(value)

    except:
        return np.nan


df["delivery_time_days"] = df["delivery_time_days"].apply(
    convert_delivery_days
)

print("Delivery time column fixed successfully!")
print(df["delivery_time_days"].head(10))
print("\nData Type:")
print(df["delivery_time_days"].dtype)

print("\nMissing Values:")
print(df["delivery_time_days"].isnull().sum())
```

# Hypothesis 1 — Discount and Sales

```python
#  Hypothesis 1 - Discount and Sales

if "discount" in df.columns and "calculated_sales" in df.columns:

    discount_sales_correlation = df["discount"].corr(
        df["calculated_sales"]
    )

    print("Hypothesis 1:")
    print("Higher discount is associated with higher sales.")
    print(
        "Correlation between discount and sales:",
        round(discount_sales_correlation, 3)
    )

    if discount_sales_correlation > 0.05:
        print("Result: Positive relationship found.")

    elif discount_sales_correlation < -0.05:
        print("Result: Negative relationship found.")

    else:
        print("Result: No strong relationship found.")

else:
    print("Required columns are not available.")
```

# Hypothesis 2 — Delivery Time and Returns

```python
# Hypothesis 2 - Delivery Time and Returns

if "delivery_time_days" in df.columns and "returned" in df.columns:

    temp_df = df.copy()

    # Convert delivery time into numeric values
    temp_df["delivery_time_days"] = pd.to_numeric(
        temp_df["delivery_time_days"],
        errors="coerce"
    )

    # Remove invalid delivery time values
    temp_df = temp_df.dropna(
        subset=["delivery_time_days", "returned"]
    )

    return_analysis = (
        temp_df.groupby("returned")["delivery_time_days"]
        .mean()
    )

    display(return_analysis)

    print("Hypothesis 2:")
    print("Orders with longer delivery time may have more returns.")

    if "Yes" in return_analysis.index and "No" in return_analysis.index:

        if return_analysis["Yes"] > return_analysis["No"]:
            print(
                "Result: Returned orders had higher average delivery time."
            )

        else:
            print(
                "Result: Returned orders did not have higher average delivery time."
            )

else:
    print("Delivery time or returned column is not available.")
```

# Hypothesis 3 — Category and Sales

```python
# Hypothesis 3 - Product Category and Sales

if "category" in df.columns and "calculated_sales" in df.columns:

    category_analysis = (
        df.groupby("category")["calculated_sales"]
        .mean()
        .sort_values(ascending=False)
    )

    display(category_analysis)

    print("Hypothesis 3:")
    print("Different product categories have different average sales.")

    print(
        "Result: Average sales vary across product categories."
    )

else:
    print("Category or sales column is not available.")
```
