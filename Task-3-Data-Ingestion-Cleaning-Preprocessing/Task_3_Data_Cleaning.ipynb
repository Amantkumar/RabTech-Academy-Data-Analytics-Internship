# Import Libraries

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display all columns
pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

# Load Dataset

```python
# Load raw CSV dataset
df = pd.read_csv("raw_dataset.csv")

# Store original dataset details
original_rows = df.shape[0]
original_columns = df.shape[1]
original_missing_values = df.isnull().sum().sum()
original_duplicates = df.duplicated().sum()

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

# Display First Five Rows

```python
# Display first five rows
display(df.head())
```

# Display Dataset Information

```python
# Display dataset information
df.info()
```

# Statistical Summary

```python
# Display statistical summary
display(df.describe(include="all"))
```

# Check Missing Values

```python
# Check missing values in each column
missing_values = df.isnull().sum()

# Calculate missing percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Create missing value summary
missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

display(missing_summary)
```

# Check Duplicate Records

```python
# Count duplicate records
duplicate_count = df.duplicated().sum()

print("Duplicate Records:", duplicate_count)
```

# Standardize Column Names

```python
# Convert column names into lowercase
df.columns = df.columns.str.lower()

# Replace spaces with underscores
df.columns = df.columns.str.replace(" ", "_")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

print("Standardized Column Names:")
print(df.columns.tolist())
```

# Remove Extra Spaces

```python
# Select text columns
text_columns = df.select_dtypes(include="object").columns

# Remove extra spaces from text columns
for column in text_columns:
    df[column] = df[column].str.strip()

print("Extra spaces removed successfully!")
```

# Correct Data Types

```python
# Convert object columns to numeric wherever possible
for column in df.columns:
    if df[column].dtype == "object":
        converted_column = pd.to_numeric(
            df[column],
            errors="ignore"
        )
        df[column] = converted_column

print("Data types corrected successfully!")
display(df.dtypes)
```

# Handle Missing Values

```python
# Handle missing values in numerical columns using median
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

# Handle missing values in text columns using mode
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(
            df[column].mode()[0]
        )

print("Missing values handled successfully!")
```

# Convert Date Columns

```python
# Detect possible date columns
date_columns = [
    column for column in df.columns
    if "date" in column or "time" in column
]

# Convert date columns into datetime format
for column in date_columns:
    df[column] = pd.to_datetime(
        df[column],
        errors="coerce"
    )

print("Date columns converted successfully!")
print(date_columns)
```

# Remove Duplicate Records

```python
# Check duplicate records
duplicate_count = df.duplicated().sum()

print("Duplicate Records:", duplicate_count)

# Remove duplicates only if duplicates are present
if duplicate_count > 0:
    df = df.drop_duplicates()
    df = df.reset_index(drop=True)
    print("Duplicate records removed!")
else:
    print("No duplicate records found. No changes required.")
```

# Detect Outliers

```python
# Detect outliers using the IQR method
numeric_columns = df.select_dtypes(include=np.number).columns

outlier_summary = {}

for column in numeric_columns:

    # Calculate Q1 and Q3
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Calculate limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Count outliers
    outlier_count = (
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ).sum()

    outlier_summary[column] = outlier_count

# Display outlier summary
outlier_df = pd.DataFrame(
    list(outlier_summary.items()),
    columns=["Column", "Outlier Count"]
)

display(outlier_df)
```

# Handle Outliers

```python
# Handle outliers using IQR clipping
for column in numeric_columns:

    # Calculate Q1 and Q3
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Calculate limits
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Clip extreme values
    df[column] = df[column].clip(
        lower=lower_limit,
        upper=upper_limit
    )

print("Outliers handled successfully!")
```

# Feature Engineering - Year and Month

```python
# Create year and month features from date columns
for column in date_columns:

    # Create year column
    df[column + "_year"] = df[column].dt.year

    # Create month column
    df[column + "_month"] = df[column].dt.month

print("Year and month features created!")
display(df.head())
```

# Calculate Sales and Profit Features

```python
# Check whether required columns are available
if all(column in df.columns for column in ["price", "quantity"]):

    # Calculate total sales
    df["calculated_sales"] = (
        df["price"] * df["quantity"]
    )

    print("Sales column created!")

# Calculate estimated profit if shipping cost is available
if all(
    column in df.columns
    for column in ["calculated_sales", "shipping_cost"]
):

    # Calculate estimated profit
    df["estimated_profit"] = (
        df["calculated_sales"] -
        df["shipping_cost"]
    )

    print("Estimated profit column created!")
```

# Before and After Comparison

```python
# Create before and after cleaning comparison
comparison = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Records"
    ],

    "Before Cleaning": [
        original_rows,
        original_columns,
        original_missing_values,
        original_duplicates
    ],

    "After Cleaning": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
})

display(comparison)
```

# Final Dataset Check

```python
# Display final dataset shape
print("Final Dataset Shape:", df.shape)

# Check missing values
print("\nMissing Values After Cleaning:")
display(df.isnull().sum())

# Check duplicate records
print(
    "\nDuplicate Records After Cleaning:",
    df.duplicated().sum()
)

# Display cleaned dataset
display(df.head(10))
```

```python
# Export cleaned dataset as clean_dataset.csv
df.to_csv(
    "clean_dataset.csv",
    index=False
)

print("clean_dataset.csv created successfully!")
```
